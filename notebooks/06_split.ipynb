```python
import sys; sys.path.append("..")
import pandas as pd
from sklearn.model_selection import train_test_split
from src.preprocess import clean_rules, add_features

raw = pd.read_csv("../data/raw/cs-training.csv", index_col=0)
df = add_features(clean_rules(raw))
target = "SeriousDlqin2yrs"

X = df.drop(columns=[target])   # features
y = df[target]                  # labels

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(X_train.shape, X_test.shape)
print(y_train.mean(), y_test.mean())
```

```python
import os
os.makedirs("../data/processed", exist_ok=True)

X_train.assign(**{target: y_train}).to_csv("../data/processed/train.csv")
X_test.assign(**{target: y_test}).to_csv("../data/processed/test.csv")
```
