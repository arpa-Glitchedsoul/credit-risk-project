```python
import sys; sys.path.append("..")
import numpy as np, pandas as pd
from sklearn.dummy import DummyClassifier
from src.evaluate import report

train = pd.read_csv("../data/processed/train.csv", index_col=0)
y = train["SeriousDlqin2yrs"]
X_dummy = np.zeros((len(y), 1))   # a dummy model ignores features

lazy = DummyClassifier(strategy="prior", random_state=42).fit(X_dummy, y)
pred = lazy.predict(X_dummy)
proba = lazy.predict_proba(X_dummy)[:, 1]
report(y, pred, proba)
```
