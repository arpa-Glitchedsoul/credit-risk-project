```python
import sys; sys.path.append("..")
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from src.train import make_logreg
from src.evaluate import report

train = pd.read_csv("../data/processed/train.csv", index_col=0)
y = train.pop("SeriousDlqin2yrs")
X = train

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
proba = cross_val_predict(make_logreg(), X, y, cv=cv, method="predict_proba")[:, 1]
pred = (proba >= 0.5).astype(int)    # default cutoff for now; Part 11 improves it

report(y, pred, proba)
```
