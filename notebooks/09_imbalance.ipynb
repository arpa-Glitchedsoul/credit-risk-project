```python
import sys; sys.path.append("..")
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from src.train import make_logreg, make_logreg_smote
from src.evaluate import report

train = pd.read_csv("../data/processed/train.csv", index_col=0)
y = train.pop("SeriousDlqin2yrs")
X = train
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rows = {}
for name, model in [("class_weight", make_logreg()), ("smote", make_logreg_smote())]:
    proba = cross_val_predict(model, X, y, cv=cv, method="predict_proba")[:, 1]
    rows[name] = report(y, (proba >= 0.5).astype(int), proba)

pd.DataFrame(rows).T.round(3)
```
