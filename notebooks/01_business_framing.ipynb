```python
# ASSUMPTIONS (illustrative, not from the dataset). We revisit these in Part 11.
LOAN_AMOUNT = 10_000          # size of a typical loan
INTEREST_PROFIT = 1_000       # profit if a good borrower repays
LOSS_GIVEN_DEFAULT = 0.60     # share of the loan lost when a borrower defaults

COST_FP = INTEREST_PROFIT                    # rejected a good borrower: lost profit
COST_FN = LOAN_AMOUNT * LOSS_GIVEN_DEFAULT   # approved a defaulter: lost money

print(f"FP cost: {COST_FP}, FN cost: {COST_FN}, ratio: {COST_FN / COST_FP:.1f}:1")
```
